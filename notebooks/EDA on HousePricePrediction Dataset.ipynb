## Data Cleaning

```python
duplicate_rows_df = df[df.duplicated()]
print("number of duplicate rows: ", duplicate_rows_df.shape)
```

```python
df = df.drop_duplicates()
df.head(5)
```

```python
# Identifying the missing or null values.
print(df.isnull().sum())
```

```python
df = df.dropna()    # Dropping the missing values.
df.count()
```

```python
print(df.isnull().sum())   # After dropping the values
```

## Data Visualization

```python
# Plot the target variable distribution
plt.hist(x=df['SalePrice'], bins=30, color='skyblue', edgecolor='black')

plt.xlabel('SalePrice')
plt.ylabel('Frequency')
plt.title('Sale Price Distribution')

plt.show()
```

```python
# Detecting outliers in key numeric features
sns.boxplot(x=df['SalePrice'])
```

```python
sns.boxplot(x=df['TotalBsmtSF'])
```

```python
sns.boxplot(x=df['LotArea'])
```

```python
plt.figure(figsize=(12, 8))

# Correlation matrix for numeric columns only
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, fmt='.2f', cmap='Pastel2', linewidths=1)

plt.title('Correlation Heatmap')
plt.show()
```

```python
# Scatter plot: house age vs price
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(df['YearBuilt'], df['SalePrice'])
ax.set_xlabel('YearBuilt')
ax.set_ylabel('SalePrice')
plt.show()
```
